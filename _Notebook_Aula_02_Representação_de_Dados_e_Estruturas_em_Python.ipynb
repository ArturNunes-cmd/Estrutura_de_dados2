```
class Paciente:
    def __init__(self, nome: str, idade: int, prioritario: bool = False):
        self.nome = nome
        self.idade = idade
        self.prioritario = prioritario

    def __repr__(self):
        tipo = "Prioritário" if self.prioritario else "Comum"
        return f"Paciente({self.nome}, {self.idade} anos, {tipo})"


class Node:
    def __init__(self, paciente: Paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def esta_vazia(self) -> bool:
        return self.inicio is None

    def tamanho(self) -> int:
        return self._tamanho

    def adicionar(self, paciente: Paciente):
        novo_no = Node(paciente)
        self._tamanho += 1

        if self.esta_vazia():
            self.inicio = novo_no
            self.fim = novo_no
            return

        if paciente.prioritario:
            if not self.inicio.paciente.prioritario:
                novo_no.proximo = self.inicio
                self.inicio = novo_no
                return

            atual = self.inicio
            while atual.proximo and atual.proximo.paciente.prioritario:
                atual = atual.proximo

            novo_no.proximo = atual.proximo
            atual.proximo = novo_no
            if novo_no.proximo is None:
                self.fim = novo_no
        else:
            self.fim.proximo = novo_no
            self.fim = novo_no

    def atender(self) -> Paciente:
        if self.esta_vazia():
            return None

        paciente_atendido = self.inicio.paciente
        self.inicio = self.inicio.proximo
        self._tamanho -= 1

        if self.inicio is None:
            self.fim = None

        return paciente_atendido

    def listar(self):
        if self.esta_vazia():
            print("\nFila vazia.")
            return

        print("\n--- PACIENTES NA FILA ---")
        atual = self.inicio
        posicao = 1
        while atual:
            print(f"{posicao}º -> {atual.paciente}")
            atual = atual.proximo
            posicao += 1


# ==========================================
# LOOP INTERATIVO (MENU)
# ==========================================
def menu():
    fila = FilaAtendimento()

    while True:
        print("\n" + "="*30)
        print(" SISTEMA DE FILA DE ATENDIMENTO")
        print("="*30)
        print("1 - Cadastrar Paciente")
        print("2 - Atender Paciente")
        print("3 - Listar Fila")
        print("4 - Ver quantidade de pacientes")
        print("0 - Sair")

        opcao = input("\nEscolha uma opção: ").strip()

        if opcao == "1":
            nome = input("Nome do paciente: ").strip()

            try:
                idade = int(input("Idade: "))
            except ValueError:
                print("⚠️ Idade inválida! Cadastro cancelado.")
                continue

            # Define prioridade automatica por idade (>60) ou pergunta manual
            if idade >= 60:
                print("-> Prioridade automática aplicada (Idade >= 60).")
                prioritario = True
            else:
                resp = input("Paciente possui outra prioridade? (s/n): ").strip().lower()
                prioritario = resp == "s"

            paciente = Paciente(nome, idade, prioritario)
            fila.adicionar(paciente)
            print(f"\n✅ {nome} adicionado(a) à fila com sucesso!")

        elif opcao == "2":
            paciente = fila.atender()
            if paciente:
                print(f"\n🔔 Atendendo agora: {paciente}")
            else:
                print("\n⚠️ Nenhum paciente na fila para atender.")

        elif opcao == "3":
            fila.listar()

        elif opcao == "4":
            print(f"\n📊 Total de pacientes aguardando: {fila.tamanho()}")

        elif opcao == "0":
            print("\nEncerrando o sistema...")
            break

        else:
            print("\n⚠️ Opção inválida! Tente novamente.")


if __name__ == "__main__":
    menu()
```
