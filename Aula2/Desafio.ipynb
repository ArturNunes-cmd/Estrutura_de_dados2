```
# Cria a classe Aluno
class Aluno:

    # Método construtor da classe
    def __init__(self, nome, idade, notas):
        self.nome = nome
        self.idade = idade
        self.notas = notas


# Cria uma lista vazia para armazenar os alunos
alunos = []


# Repete o cadastro de alunos 5 vezes
for i in range(5):

    # Mostra qual aluno está sendo cadastrado
    print(f"\nAluno {i + 1}")

    # Recebe o nome do aluno
    nome = str(input("Nome: "))

    # Recebe a idade do aluno e converte para inteiro
    idade = int(input("Idade: "))

    # Cria uma lista vazia para armazenar as notas
    notas = []

    # Repete 3 vezes para receber as notas
    for j in range(3):

        # Recebe uma nota e converte para número decimal
        nota = float(input(f"Nota {j + 1} escreva de 0 a 10: "))

        # Adiciona a nota na lista de notas
        notas.append(nota)

    # Cria um objeto da classe Aluno
    aluno = Aluno(nome, idade, notas)

    # Adiciona o aluno na lista de alunos
    alunos.append(aluno)


# Define a maior média inicialmente como 0
maior_media = 0

# Variável que vai guardar o aluno com a maior média
aluno_maior_media = None


# Mostra o título do resultado
print("\n--- Resultado ---")


# Percorre todos os alunos cadastrados
for aluno in alunos:

    # Calcula a média das 3 notas do aluno
    media = sum(aluno.notas) / 3

    # Mostra o nome do aluno e sua média com 2 casas decimais
    print(f"Aluno: {aluno.nome} | Média: {media:.2f}")

    # Verifica se a média atual é maior que a maior média encontrada
    if media > maior_media:

        # Atualiza a maior média
        maior_media = media

        # Guarda o aluno que possui a maior média
        aluno_maior_media = aluno


# Mostra o nome do aluno com a maior média
print(f"\nAluno com maior média: {aluno_maior_media.nome}")

# Mostra a maior média encontrada com 2 casas decimais
print(f"Maior média: {maior_media:.2f}")
```
