```python
class Aluno:
    def __init__(self, nome, idade, notas):
        self.nome = nome
        self.idade = idade
        self.notas = notas

alunos = []

for i in range(5):
    print(f"\nAluno {i + 1}")

    nome = str(input("Nome: "))
    idade = int(input("Idade: "))

    notas = []
    for j in range(3):
        nota = float(input(f"Nota {j + 1} escreva de 0 a 10: "))
        notas.append(nota)

    aluno = Aluno(nome, idade, notas)
    alunos.append(aluno)

maior_media = 0
aluno_maior_media = None

print("\n--- Resultado ---")

for aluno in alunos:
    media = sum(aluno.notas) / 3

    print(f"Aluno: {aluno.nome} | Média: {media:.2f}")

    if media > maior_media:
        maior_media = media
        aluno_maior_media = aluno

print(f"\nAluno com maior média: {aluno_maior_media.nome}")
print(f"Maior média: {maior_media:.2f}")
```
