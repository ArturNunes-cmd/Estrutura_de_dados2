```python
from collections import deque

fila = deque()

while True:
    arquivo = input("Digite o nome do arquivo ou 'sair' para encerrar: ")

    if arquivo.lower() == "sair":
        break

    fila.append(arquivo)

print("\nFila de impressão:")

while fila:
    arquivo = fila.popleft()
    print("Imprimindo:", arquivo)

print("Fila vazia.")
```
