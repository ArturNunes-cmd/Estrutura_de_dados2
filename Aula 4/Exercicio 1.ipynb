```python
pilha = []

while True:
    print("\n1 - Digitar")
    print("2 - Apagar")
    print("3 - Substituir")
    print("4 - Desfazer")
    print("5 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        texto = input("Digite o texto: ")
        pilha.append("Digitar: " + texto)

    elif opcao == "2":
        texto = input("Digite o que deseja apagar: ")
        pilha.append("Apagar: " + texto)

    elif opcao == "3":
        antigo = input("Digite o texto que deseja substituir: ")
        novo = input("Digite o novo texto: ")
        pilha.append("Substituir: " + antigo + " por " + novo)

    elif opcao == "4":
        if len(pilha) > 0:
            acao = pilha.pop()
            print("Desfazendo:", acao)
        else:
            print("Não há ações para desfazer.")

    elif opcao == "5":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")

    print("Pilha:", pilha)
```
