```
# Cria uma lista vazia para representar a pilha
pilha = []

# Loop principal do programa
while True:

    # Mostra o menu de opções
    print("\n1 - Digitar")
    print("2 - Apagar")
    print("3 - Substituir")
    print("4 - Desfazer")
    print("5 - Sair")

    # Recebe a opção escolhida pelo usuário
    opcao = input("Escolha uma opção: ")

    # Opção 1 - Digitar um texto
    if opcao == "1":
        texto = input("Digite o texto: ")

        # Adiciona a ação no final da pilha
        pilha.append("Digitar: " + texto)

    # Opção 2 - Apagar um texto
    elif opcao == "2":
        texto = input("Digite o que deseja apagar: ")

        # Adiciona a ação no final da pilha
        pilha.append("Apagar: " + texto)

    # Opção 3 - Substituir um texto
    elif opcao == "3":
        antigo = input("Digite o texto que deseja substituir: ")
        novo = input("Digite o novo texto: ")

        # Adiciona a ação de substituição na pilha
        pilha.append("Substituir: " + antigo + " por " + novo)

    # Opção 4 - Desfazer a última ação
    elif opcao == "4":

        # Verifica se existem ações na pilha
        if len(pilha) > 0:

            # Remove e guarda o último elemento da pilha
            acao = pilha.pop()

            # Mostra qual ação foi desfeita
            print("Desfazendo:", acao)

        # Caso a pilha esteja vazia
        else:
            print("Não há ações para desfazer.")

    # Opção 5 - Sair do programa
    elif opcao == "5":
        print("Programa encerrado.")

        # Encerra o loop
        break

    # Caso o usuário digite uma opção que não existe
    else:
        print("Opção inválida.")

    # Mostra o estado atual da pilha
    print("Pilha:", pilha)
```
