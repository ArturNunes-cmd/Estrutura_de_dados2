```
# Classe que representa um paciente
class Paciente:

    # Método construtor da classe
    def __init__(self, nome, idade, prioritario=False):
        self.nome = nome
        self.idade = idade
        self.prioritario = prioritario

    # Define como o paciente será mostrado na tela
    def __repr__(self):
        # Define o tipo do paciente
        tipo = "Prioritário" if self.prioritario else "Comum"

        # Retorna as informações do paciente
        return f"{self.nome} - {self.idade} anos - {tipo}"


# Classe que representa um nó da fila
class No:

    # Método construtor do nó
    def __init__(self, paciente):
        # Guarda o paciente dentro do nó
        self.paciente = paciente

        # Guarda o próximo nó da fila
        self.proximo = None


# Classe responsável por controlar a fila de atendimento
class FilaAtendimento:

    # Método construtor da fila
    def __init__(self):
        # Guarda o primeiro elemento da fila
        self.inicio = None

        # Guarda o último elemento da fila
        self.fim = None

        # Guarda a quantidade de pacientes
        self.tamanho = 0

    # Verifica se a fila está vazia
    def esta_vazia(self):
        return self.inicio is None

    # Adiciona um paciente na fila
    def adicionar(self, paciente):

        # Cria um novo nó para o paciente
        novo = No(paciente)

        # Verifica se a fila está vazia
        if self.esta_vazia():

            # O novo paciente será o primeiro da fila
            self.inicio = novo

            # O novo paciente também será o último
            self.fim = novo

        # Verifica se o paciente é prioritário
        elif paciente.prioritario:

            # Verifica se o primeiro paciente não é prioritário
            if not self.inicio.paciente.prioritario:

                # Coloca o novo paciente no início da fila
                novo.proximo = self.inicio
                self.inicio = novo

            else:
                # Começa a percorrer a fila pelo início
                atual = self.inicio

                # Procura o último paciente prioritário
                while atual.proximo and atual.proximo.paciente.prioritario:
                    atual = atual.proximo

                # Coloca o novo paciente depois dos prioritários
                novo.proximo = atual.proximo
                atual.proximo = novo

                # Se o novo paciente for o último da fila,
                # atualiza o fim da fila
                if novo.proximo is None:
                    self.fim = novo

        else:
            # Adiciona o paciente comum no final da fila
            self.fim.proximo = novo
            self.fim = novo

        # Aumenta a quantidade de pacientes
        self.tamanho += 1

    # Remove e retorna o primeiro paciente da fila
    def atender(self):

        # Verifica se a fila está vazia
        if self.esta_vazia():
            return None

        # Guarda o paciente que será atendido
        paciente = self.inicio.paciente

        # Move o início da fila para o próximo paciente
        self.inicio = self.inicio.proximo

        # Diminui a quantidade de pacientes
        self.tamanho -= 1

        # Se a fila ficou vazia, o fim também recebe None
        if self.inicio is None:
            self.fim = None

        # Retorna o paciente que foi atendido
        return paciente

    # Mostra todos os pacientes da fila
    def listar(self):

        # Verifica se a fila está vazia
        if self.esta_vazia():
            print("Fila vazia.")
            return

        # Começa pelo primeiro paciente
        atual = self.inicio

        # Define a posição inicial
        posicao = 1

        # Percorre todos os nós da fila
        while atual:

            # Mostra a posição e os dados do paciente
            print(f"{posicao} - {atual.paciente}")

            # Passa para o próximo paciente
            atual = atual.proximo

            # Aumenta a posição
            posicao += 1


# Função que controla o menu do sistema
def menu():

    # Cria uma fila de atendimento
    fila = FilaAtendimento()

    # Mantém o menu funcionando até o usuário escolher sair
    while True:

        # Mostra o menu principal
        print("\n=== FILA DE ATENDIMENTO ===")
        print("1 - Cadastrar paciente")
        print("2 - Atender paciente")
        print("3 - Listar fila")
        print("4 - Quantidade de pacientes")
        print("0 - Sair")

        # Recebe a opção escolhida pelo usuário
        opcao = input("Escolha uma opção: ")

        # Opção 1 - Cadastrar paciente
        if opcao == "1":

            # Recebe o nome do paciente
            nome = input("Nome: ")

            # Tenta receber a idade
            try:
                idade = int(input("Idade: "))

            # Caso o usuário digite algo que não seja número
            except ValueError:
                print("Idade inválida.")
                continue

            # Pessoas com 60 anos ou mais são prioritárias
            if idade >= 60:
                prioritario = True
                print("Paciente prioritário por idade.")

            else:
                # Pergunta se existe outra prioridade
                resposta = input("Possui outra prioridade? (s/n): ").lower()

                # Define como prioritário caso a resposta seja "s"
                prioritario = resposta == "s"

            # Cria um objeto Paciente
            paciente = Paciente(nome, idade, prioritario)

            # Adiciona o paciente na fila
            fila.adicionar(paciente)

            # Mostra mensagem de sucesso
            print("Paciente cadastrado com sucesso.")

        # Opção 2 - Atender paciente
        elif opcao == "2":

            # Remove o primeiro paciente da fila
            paciente = fila.atender()

            # Verifica se algum paciente foi encontrado
            if paciente:
                print(f"Atendendo: {paciente}")

            else:
                print("Não há pacientes na fila.")

        # Opção 3 - Listar pacientes
        elif opcao == "3":

            # Mostra todos os pacientes da fila
            fila.listar()

        # Opção 4 - Mostrar quantidade de pacientes
        elif opcao == "4":

            # Mostra o tamanho atual da fila
            print(f"Quantidade de pacientes: {fila.tamanho}")

        # Opção 0 - Sair
        elif opcao == "0":

            # Mostra mensagem de encerramento
            print("Sistema encerrado.")

            # Encerra o loop
            break

        # Caso seja digitada uma opção inexistente
        else:
            print("Opção inválida.")


# Verifica se este arquivo está sendo executado diretamente
if __name__ == "__main__":

    # Inicia o sistema pelo menu
    menu()
```
